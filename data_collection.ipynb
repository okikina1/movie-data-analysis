# Data Collection Notebook
#
# This notebook collects movie data from the TMDB API and saves it as movies_data.csv.
# The notebook has already been executed. The resulting dataset is included in the repository.
# For analysis, see analysis.ipynb.

```python
import requests
import pandas as pd
from google.colab import userdata
```

```python
# token from colab secrets and url for the TMDB API auth
my_token = userdata.get('TMDBKEY')
```

```python
headers = {
    "accept": "application/json",
    "Authorization": f"Bearer {my_token}"
}
```

```python
all_movies = []

for page in range(1, 26):   # 25 pages = 500 movies
    movies_url = f"https://api.themoviedb.org/3/movie/popular?page={page}"
    movies_response = requests.get(movies_url, headers=headers)
    movies_data = movies_response.json()

    for movie in movies_data["results"]:
        movie_id = movie["id"]

        detail_url = f"https://api.themoviedb.org/3/movie/{movie_id}"
        detail_response = requests.get(detail_url, headers=headers)
        detail_data = detail_response.json()


        genres_list = [g["name"] for g in detail_data.get("genres", [])]

        genres_string = ", ".join(genres_list)


        release_date = detail_data.get("release_date")

        if release_date:
            year = release_date[:4]
        else:
            year = None



        all_movies.append({
            "movie_id": detail_data.get("id"),
            "title": detail_data.get("title"),
            "release_date": release_date,
            "year": year,
            "genres": genres_string,
            "revenue": detail_data.get("revenue", 0),
            "budget": detail_data.get("budget", 0),
            "runtime": detail_data.get("runtime"),
            "rating": detail_data.get("vote_average")
        })

movies_df = pd.DataFrame(all_movies)

movies_df.to_csv("movies_data.csv", index=False)

print("CSV file saved successfully.")
movies_df.head()
```
